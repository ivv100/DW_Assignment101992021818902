# Assignment 3 – Sales Data Using Apache Spark

**Aim:** Take sales data and write a Spark program to calculate the number of products sold in each category.

**Data source:** Sample sales data created for this assignment for educational/demo purposes. The accompanying file is `sales_data.csv`.

**Task:** Group sales records by `category` and calculate the total number of products sold using the `quantity` column.

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum as _sum

spark = SparkSession.builder.appName("SalesByCategory").getOrCreate()
print("Spark started successfully")
```

```python
# Read the CSV sales data
df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv("sales_data.csv")
)

df.show(10, truncate=False)
```

### Spark calculation

We group the records by product category and **sum the `quantity`** field. This gives the total number of individual products sold in each category.

```python
# Calculate total number of products sold in each category
products_by_category = (
    df.groupBy("category")
      .agg(_sum("quantity").alias("products_sold"))
      .orderBy("category")
)

products_by_category.show()
```

```python
# Stop the Spark session after the work is complete
spark.stop()
```

## Conclusion

The Spark program successfully groups sales data by category and calculates the total number of products sold in every category using `groupBy()` and `sum()`.
