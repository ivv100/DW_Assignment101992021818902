```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Spark Practical ").master("local[*]").getOrCreate()

print("Spark Session Created")
```

```python
print("Version ", spark.version)
```

```python
print("Spark App Name ", spark.sparkContext.appName)
```

```python
sc = spark.sparkContext
print("Spark Context Created")
print(sc)
```

```python
data = [1,2,3,4,5]

rdd = sc.parallelize(data)

print("RDD Created Successfully")
print("RDD data ", rdd.collect())
```

```python
print(type(rdd))
```

```python
print("Number of Partitions:",rdd.getNumPartitions())
```

```python
data = list(range(1,25))
rdd = sc.parallelize(data,6)
print("RDD data",rdd.collect())
print("Number of Partitions",rdd.getNumPartitions())
```

```python
partitions = rdd.glom().collect()

for i, partition in enumerate(partitions):
  print("Partition",i+1,":",partition)
```

```python
mapped_rdd = rdd.map(lambda x: x*2)
print("Mapped RDD Data",mapped_rdd.collect())
```

```python
even_rdd=rdd.filter(lambda x: x%2==0)
print("Even RDD Data",even_rdd.collect())
```

```python
result = rdd.filter(lambda x: x<10).map(lambda x: x*3)
print("Result",result.collect())
```

```python
print("Count",rdd.count())
print("last",rdd.take(10))
print("First",rdd.first())
print("Max",rdd.max())
print("Min",rdd.min())
```

```python
total = rdd.reduce(lambda x,y: x+y)
print("Total",total)
```

```python
print("Transformation\t\tActions")
print("")
```

```python
data = [1,2,3,4,5]
rdd = sc.parallelize(data)

step1 = rdd.map(lambda x: x*2)
step2 = step1.filter(lambda x: x>5)
result = step2.collect()

print("Result",result)
```

```python
text = ["Big Data Big Data Spark"]

rdd = sc.parallelize(text)
print(rdd.collect())
```

```python
word = rdd.flatMap(lambda x: x.split())
print(word.collect())
```

```python
word_pairs = word.map(lambda x: (x,1))
print(word_pairs.collect())
```

```python
word_cnt = word_pairs.reduceByKey(lambda x,y: x+y)
print(word_cnt.collect())
```

```python
stu_data = [
    (1,"Ram","CSE",85),
    (2,"Sita","CSE",90),
    (3,"Hari","AI",70)
]

column = ["ID","Name","Dept","Marks"]

df = spark.createDataFrame(stu_data,column)

df.show()
```

```python
df.filter(df.Marks>80).show()
```

```python
df.groupBy("Dept").count().show()
```

```python
df.createOrReplaceTempView("students")

result = spark.sql(""" SELECT * FROM students""")
result.show()
```
